````python
# Required packages:
# %pip install -q pymupdf langchain-core langchain-text-splitters \
#     langchain-openai langchain-chroma chromadb pydantic tiktoken

import os
import uuid
import pymupdf

from pydantic import BaseModel, Field
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_chroma import Chroma


# ------------------------------------------------------------------
# Configuration
# ------------------------------------------------------------------

PDF_PATH = "data/2306.02707v1.pdf"
QUERY = "What is instruction tuning?"

RETRIEVAL_K = 4       # Chroma candidates sent to reranker
FINAL_K = 2           # Reranked chunks sent to answer model
CHUNK_SIZE = 300      # tokens, not characters
CHUNK_OVERLAP = 40

assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY first"
assert os.path.exists(PDF_PATH), f"PDF not found: {PDF_PATH}"


# ------------------------------------------------------------------
# 1. Read the PDF, preserving page numbers
# ------------------------------------------------------------------

documents = []

with pymupdf.open(PDF_PATH) as pdf:
    for page_index, page in enumerate(pdf):
        text = page.get_text("text").strip()

        if text:
            documents.append(
                Document(
                    page_content=text,
                    metadata={
                        "source": PDF_PATH,
                        "page": page_index + 1,
                    },
                )
            )

print(f"Loaded {len(documents)} pages")


# ------------------------------------------------------------------
# 2. Split pages into token-aware chunks
# ------------------------------------------------------------------

splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base",
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
    add_start_index=True,
)

chunks = splitter.split_documents(documents)

# Assign stable IDs for tracing retrieval and reranking.
for index, chunk in enumerate(chunks):
    chunk.metadata["chunk_id"] = f"chunk-{index:04d}"

print(f"Created {len(chunks)} chunks")


# ------------------------------------------------------------------
# 3. Embed chunks and store them in Chroma
# ------------------------------------------------------------------

embedding_model = OpenAIEmbeddings(
    model="text-embedding-3-small"
)

# Unique name prevents duplicate data when rerunning the notebook cell.
collection_name = f"pdf_rag_{uuid.uuid4().hex[:8]}"

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    collection_name=collection_name,
)

print(f"Stored {len(chunks)} chunks in Chroma collection {collection_name}")


# ------------------------------------------------------------------
# 4. Embed the query and retrieve top-k candidates from Chroma
# ------------------------------------------------------------------

retrieved = vectorstore.similarity_search_with_relevance_scores(
    QUERY,
    k=RETRIEVAL_K,
)

candidate_documents = [document for document, score in retrieved]

for document, score in retrieved:
    document.metadata["retrieval_score"] = float(score)

print("\nRetrieved candidates:")

for rank, document in enumerate(candidate_documents, start=1):
    print(
        f"{rank}. {document.metadata['chunk_id']} | "
        f"page={document.metadata['page']} | "
        f"vector_score={document.metadata['retrieval_score']:.3f}"
    )


# ------------------------------------------------------------------
# 5. Rerank the retrieved candidates with an LLM
# ------------------------------------------------------------------

class RankedChunk(BaseModel):
    chunk_id: str
    relevance_score: float = Field(ge=0, le=100)


class RerankResult(BaseModel):
    rankings: list[RankedChunk]


rerank_llm = ChatOpenAI(
    model="gpt-4.1",
    temperature=0,
    max_tokens=250,
).with_structured_output(RerankResult)

rerank_prompt = ChatPromptTemplate.from_template(
    """You are a relevance reranker.

Rank every candidate according to how directly it contains evidence
needed to answer the query.

Do not answer the query.
Return each chunk_id exactly once.
Use relevance scores from 0 to 100.

Query:
{query}

Candidates:
{candidates}
"""
)

candidate_text = "\n\n".join(
    (
        f"CHUNK_ID: {document.metadata['chunk_id']}\n"
        f"PAGE: {document.metadata['page']}\n"
        f"TEXT:\n{document.page_content}"
    )
    for document in candidate_documents
)

rerank_result = (rerank_prompt | rerank_llm).invoke(
    {
        "query": QUERY,
        "candidates": candidate_text,
    }
)

documents_by_id = {
    document.metadata["chunk_id"]: document
    for document in candidate_documents
}

valid_rankings = [
    ranking
    for ranking in rerank_result.rankings
    if ranking.chunk_id in documents_by_id
]

valid_rankings.sort(
    key=lambda ranking: ranking.relevance_score,
    reverse=True,
)

reranked_documents = []

for ranking in valid_rankings[:FINAL_K]:
    document = documents_by_id[ranking.chunk_id]
    document.metadata["rerank_score"] = ranking.relevance_score
    reranked_documents.append(document)

print("\nReranked results:")

for rank, document in enumerate(reranked_documents, start=1):
    print(
        f"{rank}. {document.metadata['chunk_id']} | "
        f"page={document.metadata['page']} | "
        f"rerank_score={document.metadata['rerank_score']:.1f}"
    )


# ------------------------------------------------------------------
# 6. Generate an answer using only the reranked evidence
# ------------------------------------------------------------------

context = "\n\n".join(
    (
        f"[Source: page {document.metadata['page']}, "
        f"{document.metadata['chunk_id']}]\n"
        f"{document.page_content}"
    )
    for document in reranked_documents
)

answer_llm = ChatOpenAI(
    model="gpt-4.1",
    temperature=0,
    max_tokens=200,
)

answer_prompt = ChatPromptTemplate.from_template(
    """Answer the question using only the supplied context.

If the context does not contain the answer, respond:
"I do not know based on the retrieved document."

Cite supporting pages using [page N].

Question:
{question}

Context:
{context}
"""
)

answer_chain = answer_prompt | answer_llm | StrOutputParser()

answer = answer_chain.invoke(
    {
        "question": QUERY,
        "context": context,
    }
)

print("\nFINAL ANSWER\n")
print(answer)
````

````python

def split_documents(
    chunk_size: int,
    knowledge_base,
    tokenizer_name= EMBEDDING_MODEL_NAME,
):
    """
    Splits documents into chunks of maximum size `chunk_size` tokens, using a specified tokenizer.

    Parameters:
    - chunk_size: The maximum number of tokens for each chunk.
    - knowledge_base: A list of LangchainDocument objects to be split.
    - tokenizer_name: (Optional) The name of the tokenizer to use. Defaults to `EMBEDDING_MODEL_NAME`.

    Returns:
    - A list of LangchainDocument objects, each representing a chunk. Duplicates are removed based on `page_content`.

    Raises:
    - ImportError: If necessary modules for tokenization are not available.
    """
    # this si different than the RCTS which is graphed
    #
    text_splitter = RecursiveCharacterTextSplitter.from_huggingface_tokenizer(
        AutoTokenizer.from_pretrained(tokenizer_name),
        chunk_size=chunk_size,
        chunk_overlap=int(chunk_size / 10),
        add_start_index=True,
        strip_whitespace=True,
    )

    docs_processed = (text_splitter.split_documents([doc]) for doc in knowledge_base)
    # Flatten list and remove duplicates more efficiently
    unique_texts = set()
    docs_processed_unique = []
    # these are langchain docs
    for doc_chunk in docs_processed:
        for doc in doc_chunk:
            if doc.page_content not in unique_texts:
                unique_texts.add(doc.page_content)
                docs_processed_unique.append(doc)

    return docs_processed_unique
````

````python
from transformers import AutoModel, AutoTokenizer  # Assuming usage of transformers library
from langchain_openai import OpenAIEmbeddings

from typing import Any
import os




def load_embedding_model(
    model_name = "openai", # "BAAI/bge-large-en-v1.5",
    device = "cuda"
) -> HuggingFaceBgeEmbeddings:
    """
    Loads an embedding model from the Hugging Face repository with specified configurations.

    Parameters:
    - model_name: The name of the model to load. Defaults to "BAAI/bge-large-en-v1.5".
    - device: The device to run the model on (e.g., 'cpu', 'cuda', 'mps'). Defaults to 'mps'.

    Returns:
    - An instance of HuggingFaceBgeEmbeddings configured with the specified model and device.

    Raises:
    - ValueError: If an unsupported device is specified.
    - OSError: If the model cannot be loaded from the Hugging Face repository.
    """
    model_kwargs = {"device": device}
    encode_kwargs = {"normalize_embeddings": True}  # For cosine similarity computation

    try:
        if model_name=="openai":
             embedding_model = OpenAIEmbeddings(model="text-embedding-3-small",api_key=api_key)
        else:
            embedding_model = HuggingFaceBgeEmbeddings(
                model_name=model_name,
                model_kwargs=model_kwargs,
                encode_kwargs=encode_kwargs,
            )
        return embedding_model
    except Exception as e:
        print(f"An error occurred while loading the model: {e}")
        raise
````

````python
from langchain_chroma import Chroma

def get_retriever(docs, embedding_model, top_k=3):
    """
    Initializes a retriever object to fetch the top_k most relevant documents based on cosine similarity.

    Parameters:
    - docs: A list of documents to be indexed and retrieved.
    - embedding_model: The embedding model to use for generating document embeddings.
    - top_k: The number of top relevant documents to retrieve. Defaults to 3.

    Returns:
    - A retriever object configured to retrieve the top_k relevant documents.

    Raises:
    - ValueError: If any input parameter is invalid.
    """
    # Example of parameter validation (optional)
    if top_k < 1:
        raise ValueError("top_k must be at least 1")

    try:
        vector_store = Chroma.from_documents(
            docs, embedding_model,
        )

        retriever = vector_store.as_retriever(k=top_k)
        # retriever.k = top_k

        return retriever
    except Exception as e:
        print(f"An error occurred while initializing the retriever: {e}")
        raise
````

````python
def retrieve_context(query, retriever):
    """
    Retrieves and reranks documents relevant to a given query.

    Parameters:
    - query: The search query as a string.
    - retriever: An instance of a Retriever class used to fetch initial documents.

    Returns:
    - A list of reranked documents deemed relevant to the query.

    """
    retrieved_docs = retriever.invoke(query)

    return retrieved_docs
````
